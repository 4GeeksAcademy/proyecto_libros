```python
import pandas as pd
import requests
import time
import os

def get_google_books_info(title):
    try:
        url = f"https://www.googleapis.com/books/v1/volumes?q=intitle:{title}"
        response = requests.get(url)
        if response.status_code == 200:
            data = response.json()
            if "items" in data and len(data["items"]) > 0:
                volume = data["items"][0]["volumeInfo"]
                return {
                    "rating": volume.get("averageRating"),
                    "numRatings": volume.get("ratingsCount"),
                    "coverImg": volume.get("imageLinks", {}).get("thumbnail"),
                }
    except:
        pass
    return {}

def get_openlibrary_info(title):
    try:
        url = f"https://openlibrary.org/search.json?title={title}"
        res = requests.get(url)
        if res.status_code == 200:
            data = res.json()
            if data['docs']:
                doc = data['docs'][0]
                return {
                    "setting": ", ".join(doc.get("place", [])) if "place" in doc else None,
                    "characters": ", ".join(doc.get("person", [])) if "person" in doc else None,
                    "awards": ", ".join(doc.get("awards", [])) if "awards" in doc else None,
                }
    except:
        pass
    return {}

def completar_info_por_bloques(df, bloque_size=10000, carpeta_guardado="../data/full_data"):
    if not os.path.exists(carpeta_guardado):
        os.makedirs(carpeta_guardado)

    total = len(df)
    bloques = (total // bloque_size) + 1

    for i in range(bloques):
        inicio = i * bloque_size
        fin = min((i + 1) * bloque_size, total)
        bloque = df.iloc[inicio:fin].copy()

        print(f"\nProcesando bloque {i + 1}/{bloques} - Libros {inicio} a {fin - 1}")

        procesados = 0
        for idx in bloque.index:
            row = bloque.loc[idx]
            update_needed = (
                pd.isna(row['rating']) or
                pd.isna(row['numRatings']) or
                pd.isna(row['coverImg']) or
                pd.isna(row['setting']) or
                pd.isna(row['characters']) or
                pd.isna(row['awards'])
            )

            if not update_needed:
                continue

            print(f"📚 {idx}: Buscando '{row['title']}'")

            info_google = get_google_books_info(row['title'])
            info_openlib = get_openlibrary_info(row['title'])

            if pd.isna(row['rating']) and info_google.get('rating') is not None:
                bloque.at[idx, 'rating'] = info_google['rating']
            if pd.isna(row['numRatings']) and info_google.get('numRatings') is not None:
                bloque.at[idx, 'numRatings'] = info_google['numRatings']
            if pd.isna(row['coverImg']) and info_google.get('coverImg') is not None:
                bloque.at[idx, 'coverImg'] = info_google['coverImg']

            if pd.isna(row['setting']) and info_openlib.get('setting') is not None:
                bloque.at[idx, 'setting'] = info_openlib['setting']
            if pd.isna(row['characters']) and info_openlib.get('characters') is not None:
                bloque.at[idx, 'characters'] = info_openlib['characters']
            if pd.isna(row['awards']) and info_openlib.get('awards') is not None:
                bloque.at[idx, 'awards'] = info_openlib['awards']

            procesados += 1
            time.sleep(1.1)

        output_path = os.path.join(carpeta_guardado, f"libros_completados_bloque_{i+1}.csv")
        bloque.to_csv(output_path, index=False)
        print(f"Guardado bloque {i+1} en: {output_path} ({procesados} libros actualizados)")
```

```python
completar_info_por_bloques(df, bloque_size=10000)
```

```python
import pandas as pd
import os
import glob

def unir_bloques_csv(carpeta_bloques, archivo_salida):
    # Buscar todos los archivos que empiezan con libros_completados_bloque_ y terminan en .csv
    archivos = sorted(glob.glob(os.path.join(carpeta_bloques, "libros_completados_bloque_*.csv")))

    print(f"🔍 Encontrados {len(archivos)} archivos:")
    for f in archivos:
        print(f" - {os.path.basename(f)}")

    # Leer y unir todos los archivos
    dfs = [pd.read_csv(f) for f in archivos]
    df_completo = pd.concat(dfs, ignore_index=True)

    # Guardar el resultado final
    df_completo.to_csv(archivo_salida, index=False)
    print(f"\nArchivo combinado guardado en: {archivo_salida}")

    return df_completo
```

```python
# Ruta donde están los bloques
carpeta_bloques = "../data/full_data"

# Ruta de salida
archivo_salida = "../data/full_data/libros_completados_final.csv"

# Ejecutar la unión
df_final = unir_bloques_csv(carpeta_bloques, archivo_salida)
```
